```python
data = {
    "Name": ["Anu", "Bala", "Charan", "Divya", "Esha", "Farhan"],
    "Department": ["CSE", "ECE", "CSE", "IT", "ECE", "IT"],
    "Marks": [85, 72, 90, 65, 78, 88],
    "Attendance": [95, 80, 92, 70, 85, 90]
}

df = pd.DataFrame(data)

print(df)
```
